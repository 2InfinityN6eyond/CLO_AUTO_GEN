```python
# Test clo conversion

import svgpathtools as svgpath
import shutil

SEWFACTORY_SEWING_PATTERN_DIR = "sewfactory__01"

garment_path_list = glob(os.path.join(".", "sewfactory", "*"))
for garment_path in tqdm(garment_path_list) :
    
    garment_name_list = []
    garment_name = []
    for identifier in os.path.basename(garment_path).split('_') :
        garment_name.append(identifier)
        if is_garment_id(identifier) :
            garment_name_list.append("_".join(garment_name))
            garment_name = []


    
    if len(garment_name_list) == 2 :
        NEW_GARMENT_PATH = os.path.join(
                SEWFACTORY_SEWING_PATTERN_DIR,
                f"{garment_name_list[0]}__01__{garment_name_list[1]}__01"
        )
    else : 
        NEW_GARMENT_PATH = os.path.join(
            SEWFACTORY_SEWING_PATTERN_DIR,
            f"{os.path.basename(garment_path)}__01"
        )
    os.makedirs(NEW_GARMENT_PATH, exist_ok=True)
    
    
    
    
    spec_path_list = glob(os.path.join(garment_path, "*specification.json"))
    for spec_path in spec_path_list :
        garment_id = "_".join(os.path.basename(spec_path).split("_")[:-1])
        garment_category = "_".join(os.path.basename(spec_path).split("_")[:-2])
        arrangement_map = garment_category_arrangement_map_map[garment_category]
        
        sewing_pattern = read_sewfactory_spec(spec_path)
        
        fabric_uuid = get_unique_random_string(set())
        random_hexa_code = generate_random_hex()
        
        clo_format_dict = {
            "FabricList": [],
            "GradingRuleTableList": [],
            "PatternList": [],
            "SymmetricDataList": [],
            "InstanceDataList": [],
            "SeamLinePairGroupList": [],
        }

        for panel_name, svg_panel in sewing_pattern.panel_dict.items() :
            svg_panel.approximate_quadratic_bezier_with_cubic_bezier(VIS = False)
            # svg_panel.approximate_arc_with_cubic_bezier(VIS = False)
            svg_panel.scale(8.5)
            svg_panel.mirror_vertical()

            if panel_name in ['lbsleeve', 'lfsleeve'] :
                sewing_pattern.panel_dict[panel_name].rotate_clockwise(-90)
            if panel_name in ['rbsleeve', 'rfsleeve'] :
                sewing_pattern.panel_dict[panel_name].rotate_clockwise(90)
            
            # if "back" in panel_name :
            #     sewing_pattern.mirror_panel_horizontally(panel_name)
            if sewing_pattern.panel_dict[panel_name].is_clockwise() != True :
                sewing_pattern.reverse_panel_path(panel_name)
            # if "back" in panel_name :
            #     sewing_pattern.reverse_panel_path(panel_name)
            
            panel_batch_point = arrangement_map[panel_name]

            panel_data_json = {
                "Name": f"{garment_id}_{panel_name}", # garment_id 추가함.
                "fGrainlineAngle": 0.0,
                "ID": f"{garment_id}_{panel_name}", # garment_id 추가함.
                "strSuperImposeSide": "None",
                "CurrentFabricUUID": fabric_uuid,
                "IsClosed": False,
                "InternalLineList": [],
                "ButtonHeadList": [],
                "ButtonHoleList": [],
                "AnnotationList": [],
                "NotchList": [],
                "IsHalfSymmetric": False,
                "ShapeInfo": {
                    "IsSlashed": False,
                    "LineList": [],
                    "HalfSymmetryPointIDMap": {},
                    "HalfSymmetryLineIDMap": {}
                },
                "ArrangementPointDataMap": {
                        "PointName": panel_batch_point,
                        "fOffSetX": 0.0,
                        "fOffSetY": 0.05,
                        "fAngle": 180.0
                }
            }
            # InstanceDataList에 json에 추가
            instance_data = {
                "OriginPatternID": panel_name,  # 패널 이름을 OriginPatternID로 사용
                "InstancePatternIDArray": []   # 빈 리스트로 추가
            }
            
            clo_format_dict["InstanceDataList"].append(instance_data)
            
            # coordinate_id_map = {}
            point_data_list = []
            point_idx = 0
            for edge_idx, edge in enumerate(svg_panel.svg_path):
                line_id = f"{panel_name}_edge_{edge_idx}"
                point_list = []
                
                if isinstance(edge, svgpath.CubicBezier):
                    point_type_list = ["Straight", "Bezier Curve", "Bezier Curve", "Straight"]
                    attribute_list = ["start", "control1", "control2", "end"]
                elif isinstance(edge, svgpath.Line):
                    point_type_list = ["Straight", "Straight"]
                    attribute_list = ["start", "end"]
                else :
                    raise ValueError(f"Unknown edge type: {type(edge)}")
                
                point_idx -= 1
                for point_type, attribute in zip(point_type_list, attribute_list) :
                    x = round(getattr(edge, attribute).real, 5)
                    y = round(getattr(edge, attribute).imag, 5)
                    
                    point_idx += 1
                    point_id = f"{panel_name}_point_{point_idx}"
                    point_data = {
                        "ID": point_id, 
                        "PointType": point_type,
                        "Position": {"x": x, "y": y},
                        "GradingRuleID": -1
                    }
                    point_data_list.append(point_data)
                    point_list.append(point_data)
                    
                line = {"ID": line_id, "PointList": point_list}
                panel_data_json["ShapeInfo"]["LineList"].append(line)
            
            # set information of last point matches first point
            panel_data_json["ShapeInfo"]["LineList"][-1]["PointList"][-1]["ID"]  = point_data_list[0]["ID"]
            panel_data_json["ShapeInfo"]["LineList"][-1]["PointList"][-1]["Position"] = point_data_list[0]["Position"]
            
            clo_format_dict["PatternList"].append(panel_data_json)
        # sewing_pattern.draw(invert_yaxis=False)


        clo_format_dict["SeamLinePairGroupList"] = []
        fabric_group = {
            "FabricName" : "Default Fabric",
            "FabricType" : "None",
            "FabricContent" : "None",
            "strBaseColorHexCode" : random_hexa_code,
            "FabricUUID" : fabric_uuid
            }
        # SeamLinePairGroupList 생성 루프

        for stch_idx, stch_info in sewing_pattern.stitch_dict.items() :
        # for idx, value in enumerate(stitch_dict.values()):
            # "Name" 생성
            group_name = f"{garment_id}_SeamLineGroup_{stch_idx}"

            # 정보 저장
            first_panel = stch_info.panel_0
            first_edge  = stch_info.edge_0
            first_shape_id = f"{garment_id}_{first_panel}"
            fstt = sewing_pattern.panel_dict[first_panel].normalized_edge_stt(first_edge)
            fend = sewing_pattern.panel_dict[first_panel].normalized_edge_end(first_edge)
            
            second_panel = stch_info.panel_1
            second_edge = stch_info.edge_1 
            second_shape_id = f"{garment_id}_{second_panel}"
            sstt = sewing_pattern.panel_dict[second_panel].normalized_edge_stt(second_edge)
            send = sewing_pattern.panel_dict[second_panel].normalized_edge_end(second_edge)
            
            first_direction = True
            second_direction = False
            
            # direction 결정 뒤에 정해 놓는 fstart, fend
            if first_direction == False:
                first_fstart, first_fend = fend, fstt
            else:
                first_fstart, first_fend = fstt, fend        
                
            if second_direction == False:
                second_fstart, second_fend = send, sstt
            else:
                second_fstart, second_fend = sstt, send

            # JSON 구조 생성
            seam_line_pair_group = {
                "Name": group_name,
                "bIsTurned": False,
                "PairList": [
                    {
                        "First": {
                            "ShapeID": first_shape_id,
                            "LengthParam": {
                                "fStart": first_fstart,
                                "fEnd": first_fend
                            },
                            "Direction": first_direction
                        },
                        "Second": {
                            "ShapeID": second_shape_id,
                            "LengthParam": {
                                "fStart": second_fstart,
                                "fEnd": second_fend
                            },
                            "Direction": second_direction 
                        }
                    }
                ],
                "FoldData": {
                    "iAngle": 180,
                    "iStrength": 5
                }
            }

            # SeamLinePairGroupList에 추가
            clo_format_dict["SeamLinePairGroupList"].append(seam_line_pair_group)

        clo_format_dict["FabricList"].append(fabric_group)    

        spec_name = os.path.basename(spec_path)
        
        save_path = os.path.join(
            NEW_GARMENT_PATH,
            spec_name.replace("_specification.json", "__01__clo.json")
        )
        try :
            shutil.copy(
                spec_path,
                os.path.join(NEW_GARMENT_PATH, os.path.basename(spec_path))
            )
        except Exception as e:
            print(f"Error copying file: {e}")
        
        with open(save_path, "w") as f:
            json.dump(clo_format_dict, f, indent=4)

    # combine top and bottom clo json if it is the case
    
    if len(garment_name_list) == 2 :
        with open(
            os.path.join(NEW_GARMENT_PATH, f"{garment_name_list[0]}__01__clo.json"),
            "r"
        ) as f:
            top_garment_json = json.load(f)
        with open(
            os.path.join(NEW_GARMENT_PATH, f"{garment_name_list[1]}__01__clo.json"),
            "r"
        ) as f:
            bottom_garment_json = json.load(f)
        
        combined_clo_json = combine_top_bottom_clo_json(
            top_garment_json,
            bottom_garment_json
        )
        
        with open(
            os.path.join(
                NEW_GARMENT_PATH,
                f"{garment_name_list[0]}__01__{garment_name_list[1]}__01__clo.json"
            ),
            "w"
        ) as f:
            json.dump(combined_clo_json, f, indent=4)
            
```

```python
# Visualize CLO sewing pattern

import math
from dataclasses import dataclass
from pprint import pprint

import matplotlib.pyplot as plt
from matplotlib.patches import FancyArrowPatch

@dataclass
class Stitch :
    shape_id : str = None
    start : float = None
    end : float = None
    direction : bool = None

@dataclass
class SeamGroup :
    name : str = None
    stitch_list : list[Stitch] = None
    

def vis_clo_sewing_pattern(clo_path, FIGLEN = 5) :
    with open(clo_path, "r") as f:
        json_data = json.load(f)

    sewing_pattern = {}
    for pattern in json_data["PatternList"] :
        svg_path = []
        
        line_list = pattern["ShapeInfo"]["LineList"]
        for line in line_list :
            point_list = line["PointList"]
            if len(point_list) == 2 : # Straight Line
                svg_path.append(svgpath.Line(
                    complex(point_list[0]["Position"]['x'], point_list[0]["Position"]["y"]),
                    complex(point_list[1]["Position"]['x'], point_list[1]["Position"]["y"])
                ))
            elif len(point_list) == 4 : # Cubic Bezier Curve
                svg_path.append(svgpath.CubicBezier(
                    complex(point_list[0]["Position"]['x'], point_list[0]["Position"]["y"]),
                    complex(point_list[1]["Position"]['x'], point_list[1]["Position"]["y"]),
                    complex(point_list[2]["Position"]['x'], point_list[2]["Position"]["y"]),
                    complex(point_list[3]["Position"]['x'], point_list[3]["Position"]["y"])
                ))

        sewing_pattern[pattern["Name"]] = svgpath.Path(*svg_path)

    seam_group_info = {}
    for seam_line_pair_group in json_data['SeamLinePairGroupList'] :
        for pair in seam_line_pair_group['PairList'] :
            seam_group_info[
                int(seam_line_pair_group['Name'].split('_')[-1])
            ] = SeamGroup(
                name = seam_line_pair_group['Name'],
                stitch_list = [
                    Stitch(
                        shape_id = pair['First']['ShapeID'],
                        start = pair['First']['LengthParam']['fStart'],
                        end = pair['First']['LengthParam']['fEnd'],
                        direction = pair['First']['Direction']
                    ),
                    Stitch(
                        shape_id = pair['Second']['ShapeID'],
                        start = pair['Second']['LengthParam']['fStart'],
                        end = pair['Second']['LengthParam']['fEnd'],
                        direction = pair['Second']['Direction']
                    )
                ]
            )

    panel_count = len(json_data['PatternList'])
    NCOLs = int(panel_count ** 0.5)
    NROWs = math.ceil(panel_count / NCOLs)
    plt.figure(figsize=(FIGLEN * NCOLs, FIGLEN * NROWs))
    
    t_arr = np.linspace(0, 1, 100)
    for panel_idx, (panel_name, path) in enumerate(sewing_pattern.items()) :
        for idx, line in enumerate(path) :
            #plt.scatter(line.point(0).real, line.point(0).imag, c='red')
            ax = plt.subplot(NROWs, NCOLs, panel_idx + 1)
            ax.text(line.point(0).real, line.point(0).imag, f'{idx}', fontsize=6)
            # plt.arrow(
            #     line.point(0).real, line.point(0).imag,
            #     line.point(1).real - line.point(0).real, line.point(1).imag - line.point(0).imag,
            #     head_width=16, head_length=16
            # )
            
            x_arr = list(map(lambda t : line.point(t).real, t_arr))
            y_arr = list(map(lambda t : line.point(t).imag, t_arr))
            ax.plot(x_arr, y_arr)
        
        for seam_group in seam_group_info.values() :
            
            corresponding_stitch = None
            for stitch in seam_group.stitch_list :
                if panel_name in stitch.shape_id :
                    corresponding_stitch = stitch
                    break
            if corresponding_stitch :
                start = path.point(corresponding_stitch.start   )
                end = path.point(corresponding_stitch.end)
                
                if corresponding_stitch.direction == False :
                    # start, end = end, start
                    pass
                    
                ax.add_patch(
                    FancyArrowPatch(
                        [start.real, start.imag],
                        [end.real, end.imag],
                        arrowstyle='-|>',
                        mutation_scale=15,
                        shrinkA=0,
                        shrinkB=0,
                        
                    )
                )
                
                ax.text(
                    (start.real + end.real) / 2,
                    (start.imag + end.imag) / 2,
                    f'{seam_group.name.split("_")[-1]}{"+" if corresponding_stitch.direction else "-" }',
                    fontsize=9,
                )

        plt.title(" ".join(panel_name.split("_")[-2:]))
        plt.axis('equal')
    plt.show()
        
```
