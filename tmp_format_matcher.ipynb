# Naming Rule

- GARMENT_VERSION 
    - Draping Rule 등 GCD/SewFactory => CLO json 포맷으로 변환 알고리즘 버전 - 변환 과정에서 Draping 방식 등이 바뀔 수 있음.
    - GARMENT_VERSION 은 데이터셋 루트 폴더에만 명시하고, 각각 Garment Directory 혹은 Garment 당 파일 등에는 전혀 명시하지 말자.

### SeamDress[GarmentCodeData]
- 각각의 Garment 별로
    - 하나의 garment 만 있을 때
        - rand_{GARMENT}
    - 두개의 garment 로 구성
        - rand_{GARMENT_ID}__rand_{GARMENT_ID}

### SeamDress[SewFactory]

```python
import os, sys
from glob import glob
from pathlib import Path

def check_legacy_format(garment_dir) :
    scene_name =  os.path.basename(garment_dir)

    if len(scene_name.split("__")) > 2 :
        return True
    elif len(scene_name.split("__")) == 2 :
        for part in scene_name.split("__") :
            if len(part.split("_")) < 2 :
                return True

    return False
```
